# Stemming :
## technique to Standardize text

```python
from nltk.stem import PorterStemmer

ps = PorterStemmer()

connect_tokens = ["Connecting", "connected","connectivity","connect","connects"]

for t in connect_tokens:
   print(f"{t:<15}: {ps.stem(t)} ") 
```

```python
learn_tokens = ['learned',"learning","learn","learns","learner","learners"]


def my_stemmer(list):
    for t in list:
        
        print(f"{t:<15}: {ps.stem(t)} ") 
my_stemmer(learn_tokens)
```

```python
likes_tokens =['likes','better','worse']

my_stemmer(likes_tokens)
```

# Lemmatization
### word is reduced to a MeaningFul base form
### Shopisticated : load pre-defined database ( dictionary)

```python
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer_obj = WordNetLemmatizer()
```

```python
for t in connect_tokens:
    print(t ,":",lemmatizer_obj.lemmatize(t))
```

```python
def wordNetLemmatizer_fun(list):
    for t in list:
        print(t ,":",lemmatizer_obj.lemmatize(t))
```

```python
wordNetLemmatizer_fun(learn_tokens)
```

```python
wordNetLemmatizer_fun(likes_tokens)
```
