```python
import nltk

from nltk.tokenize import word_tokenize , sent_tokenize
from nltk.stem import PorterStemmer , WordNetLemmatizer
from nltk.corpus import stopwords

import re
import pandas as pd
```

```python
data =pd.read_csv('tripadvisor_hotel_reviews.csv')
```

```python
data.info()
```

```python
data.head()
```

```python
data['Review'][0]
```

### Lower casing the reviews

```python
data["review_lowercase"] = data['Review'].str.lower()
```

```python
data.head()
```

### remove stopwords

```python
en_stopwords = stopwords.words("english")
en_stopwords.remove("not")
```

```python
data['review_no_stowords'] = data["review_lowercase"].apply(lambda x : ' '.join([word for word in x.split() if word not in en_stopwords]))
data['review_no_stowords'][0]
```

### removing punctuation,  remove * symbool with "star"  --- it may add meaning

```python
data['review_no_stoword_no_punct'] = data.apply(lambda x : re.sub(r'[*]',"star",x["review_no_stowords"]),axis=1)
```

```python
data.head()
```

```python
# data["review_no_stoword_no_punct"] =data["review_no_stoword_no_punct"].apply(lambda x : re.sub(r"([^\w\s])","",x))  faster btw
data["review_no_stoword_no_punct"] =data.apply(lambda x : re.sub(r"([^\w\s])","",x["review_no_stoword_no_punct"]),axis=1)
```

```python
data.head()
```

### Tokenization

```python
data["tokenized"] =data.apply(lambda x: word_tokenize(x["review_no_stoword_no_punct"]),axis =1)
data["tokenized"][0]
```

### Stemming

```python
ps =PorterStemmer()
```

```python
data["stemmed"] =data["tokenized"].apply(lambda x: [ps.stem(token) for token in x] )
```

```python
data.head()
```

### Lemmatization

```python
lemmatizer = WordNetLemmatizer()
data["lemmatized"] = data["tokenized"].apply(lambda tokenized_list : [lemmatizer.lemmatize(token) for token in tokenized_list])
data["lemmatized"][0]
```

### Ngrams
- format the big list contaion all the lemmatized word of this data sets (combined all reviews)

```python
token_clean = sum(data["lemmatized"],[])
```

```python
unigrams =(pd.Series(nltk.ngrams(token_clean,1))).value_counts()
print(unigrams)
```

```python
bigrams =(pd.Series(nltk.ngrams(token_clean,2))).value_counts()
print(bigrams)
```
