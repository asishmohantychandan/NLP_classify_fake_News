# Lowering text

```python
sentence = "Her cat's name is Luna"
lower_sentence = sentence.lower()
print(lower_sentence)
```

```python
sentence_list = ['Could you pass me the TV remote?', 
                 'It is IMPOSSIBLE to find this hotel', 
                 'Want to go for dinner on Tuesday?']

lower_sentence_list = [x.lower() for x in sentence_list]
print(lower_sentence_list)
```

# Removing Stopwords : 
- and  , of , a , to   --->  these words don't add much meaning
- NLTK library has pre-defined stopwords

```python
import nltk 
nltk.download('stopwords')
from nltk.corpus  import stopwords
```

```python
en_stopwords =  stopwords.words('english')
print(en_stopwords)
```

```python
len(en_stopwords)
```

```python
sentence = "it was too far to go to the shop and he did not want her to walk"
sentence_with_no_stowords = " ".join([word for word in sentence.split() if word not in en_stopwords])
print(sentence_with_no_stowords)
```

```python
en_stopwords.remove("did")
en_stopwords.remove("not")
en_stopwords.append("go")
```

```python
sentence_no_stowords_custom = ' '.join([ x for x in sentence.split() if x not in en_stopwords])
print(sentence_no_stowords_custom)
```

# Asish?    -- meaning letter h is optional

```python
customer_reviews = ['sam was a great help to me in the store', 
                    'the cashier was very rude to me, I think her name was eleanor', 
                    'amazing work from sadeen!', 
                    'sarah was able to help me find the items i needed quickly', 
                    'lucy is such a great addition to the team', 
                    'great service from sara she found me what i wanted'
                   ]


sarahs_reviews = []
pattern_to_find =r"sarah?"
for string in customer_reviews:
    if (re.search(pattern_to_find, string)):
        sarahs_reviews.append(string)
print(sarahs_reviews)
```

#  r"^a"   --> starts with letter 'a'

```python
a_reviews =[]
pattern_to_find =r"^a"
for x in customer_reviews:
    if re.search(pattern_to_find, x):
        a_reviews.append(x)
print(a_reviews)
```

# r"y$"  ----> end s with letter y

```python
ends_with_y_reviews =[]
pattern_to_find =r"y$"
for x in customer_reviews:
    if re.search(pattern_to_find, x):
        ends_with_y_reviews.append(x)
print(ends_with_y_reviews)
```

# r"(need|want)ed" ---> match either needed or wanted

```python
needed_wanted_reviews=[]
pattern_to_find =r"(need|want)ed"
for x in customer_reviews:
    if re.search(pattern_to_find, x):
        needed_wanted_reviews.append(x)
print(needed_wanted_reviews)
```

## removing punctuations from reviews
## find anything that is not a word character and not a space --> (puntucation)
# r"[^\w\s]"
#  [] set of characters we want to match
#  ^ meaning "not" a sit is inside [] . otherwise ^ --> start of text
#  \w  word characters
#  \s white space

```python
no_punct_reviews = []
pattern_to_find = r"[^\w\s]"
for x in customer_reviews:
    no_punct_string =  re.sub(pattern_to_find,"###", x)   #  replce with empty space
    no_punct_reviews.append(no_punct_string)
print(no_punct_reviews)
```
